## Exploratory Data Analysis (EDA)

EDA (Exploratory Data Analysis) is the process of understanding a dataset before applying machine learning. It helps us identify the data structure, missing values, data types, and basic statistics.

## Import Libraries

Import the required libraries for data analysis.

```python
import pandas as pd #use for data manipulation and analysis
import numpy as np #use for numerical computations
```

## Load Dataset

Load the Titanic dataset into a pandas DataFrame.

```python
df = pd.read_csv("train.csv")
```

## Preview Dataset

Display the first five rows of the dataset.

```python
df.head() #to see the first 5 rows of the dataset
```

## Dataset Information

View the dataset structure, data types, and missing values.

```python
df.info() #to see the information about the dataset
```

## Statistical Summary

Generate summary statistics for numerical columns.

```python
df.describe() #to see the statistical information about the dataset
```

## Dataset Shape

Check the number of rows and columns.

```python
df.shape #no of rows and columns in the dataset
```

## Column Names

Display all column names.

```python
df.columns # to see the column names of the dataset
```

## Data Types

Identify numerical and categorical columns.

```python
df.dtypes #to see the data types of the columns in the dataset
```

```python
numerical=df.select_dtypes(include=['int64','float64']).columns #to see the columns with numerical data types
categorical=df.select_dtypes(include=['object']).columns #to see the columns with categorical data types

print("Numerical Columns: ",numerical)
print("Categorical Columns: ",categorical)
```

## Missing Values

Count missing values in each column.

```python
df.isnull().sum() #to see the number of missing values in each column
```

## Data Story

The Titanic dataset contains passenger information and their survival status. It has 891 rows and 12 columns. Some columns, such as Age and Cabin, contain missing values. The dataset includes both numerical and categorical features. This analysis provides a basic understanding of the data before preprocessing and machine learning.
